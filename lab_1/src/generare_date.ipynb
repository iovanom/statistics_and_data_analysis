```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Setăm seed pentru reproducibilitate
np.random.seed(42)

# Numărul de observații
N = 100
```

```python
# Generăm datele
data = {
    'ID': range(1, N + 1),
    
    # Sex: 50-50 distribuție
    'Sex': np.random.choice(['Masculin', 'Feminin'], size=N, p=[0.5, 0.5]),
    
    # Vârstă: între 18 și 75 ani, distribuție normală centrată pe 40
    'Varsta': np.random.normal(loc=40, scale=15, size=N).astype(int),
    
    # Tip localitate: 60% urban, 40% rural
    'Tip_Localitate': np.random.choice(['Urban', 'Rural'], size=N, p=[0.6, 0.4]),
    
    # Număr pași zilnic: între 2000 și 15000, distribuție normală
    'Numar_Pasi_Zilnic': np.random.normal(loc=7500, scale=2500, size=N).astype(int),
    
    # Tip ocupație
    'Tip_Ocupatie': np.random.choice(
        ['Sedentar', 'Moderat_Activ', 'Activ', 'Student', 'Pensionar', 'Casnic/a'],
        size=N,
        p=[0.25, 0.25, 0.20, 0.15, 0.10, 0.05]
    ),
    
    # Timp liber activ: între 0 și 20 ore/săptămână
    'Timp_Liber_Activ': np.random.exponential(scale=3, size=N).round(1),
}
```

```python
# Creăm DataFrame
df = pd.DataFrame(data)

# Ajustări pentru a face datele mai realiste
# Limitează vârsta între 18 și 70 
df['Varsta'] = df['Varsta'].clip(18, 70)

# Limitează numărul de pași între 1500 și 20000
df['Numar_Pasi_Zilnic'] = df['Numar_Pasi_Zilnic'].clip(1500, 20000)

# Limitează timpul liber activ între 0 și 25 ore
df['Timp_Liber_Activ'] = df['Timp_Liber_Activ'].clip(0, 25)
```

```python
# Salvăm datele
df.to_csv('../dataset/date_activitate_fizica.csv', index=False, encoding='utf-8-sig')
```

```python
# Afișăm primele rânduri și statistici descriptive
print("Primele 10 rânduri:")
print(df.head(10))
print("\n" + "="*80 + "\n")
print("Statistici descriptive:")
print(df.describe())
print("\n" + "="*80 + "\n")
print("Distribuția variabilelor categoriale:")
print("\nSex:")
print(df['Sex'].value_counts())
print("\nTip Localitate:")
print(df['Tip_Localitate'].value_counts())
print("\nTip Ocupație:")
print(df['Tip_Ocupatie'].value_counts())
```

```python
# Generăm 24 de luni (2 ani)
n_luni = 24
date_range = pd.date_range(start='2022-01-01', periods=n_luni, freq='MS')  # MS = Month Start

# Componenta de trend (creștere liniară ușoară)
trend = 7000 + 30 * np.arange(n_luni)  # crește cu ~30 pași/lună

# Componenta sezonieră (sinusoidală, cu vârf vara și minim iarna)
# Perioada = 12 luni
sezon = 400 * np.sin(2 * np.pi * np.arange(n_luni) / 12 - np.pi/2)

# Componenta aleatoare (zgomot)
zgomot = np.random.normal(0, 150, n_luni)

# Serie finală
numar_pasi_mediu = trend + sezon + zgomot

# Creăm DataFrame
serie_dinamica = pd.DataFrame({
    'Data': date_range,
    'Numar_Pasi_Mediu': numar_pasi_mediu.round(2)
})

# Adăugăm și o coloană cu luna și anul pentru afișare
serie_dinamica['Luna_An'] = serie_dinamica['Data'].dt.strftime('%Y-%m')

print(serie_dinamica)
```

```python
# Salvăm în CSV (opțional)
serie_dinamica.to_csv('../dataset/serie_dinamica_pasi.csv', index=False)
```
