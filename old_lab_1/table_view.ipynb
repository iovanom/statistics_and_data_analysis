```python
import pandas as pd
```

```python
cols_map = {
    "ID": "ID",
    "Varsta": "Vârsta",
    "Sex": "Sex",
    "Tip_Localitate": "Localitate",
    "Tip_Cafea_Preferat": "Tip Cafea",
    "Numar_Cani_Pe_Zi": "Cani/zi",
    "Ora_Consum_Dominanta": "Perioada",
    "Buget_Luna_MDL": "Buget (MDL)",
    "Loc_Achizitie": "Loc",
}
df = pd.read_csv("./dataset/date_cafea_consum.csv")
df.reset_index()
df = df.drop("ID", axis=1)
df["Sex"] = df["Sex"].apply(lambda v: v[0])
df = df.rename(columns=cols_map)
print(df.to_latex(escape=True, index=False, longtable=True))
```
