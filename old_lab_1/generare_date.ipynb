```python
import pandas as pd
import numpy as np
from datetime import datetime, timedelta
```

```python
# Set seed pentru reproducibilitate
np.random.seed(42)

# Număr de înregistrări
n = 100
```

```python
# Generare date sintetice
data = {
    'ID': range(1, n+1),
    
    # Caracteristici demografice
    'Varsta': np.random.randint(18, 70, n),
    'Sex': np.random.choice(['Masculin', 'Feminin'], n, p=[0.48, 0.52]),
    'Tip_Localitate': np.random.choice(['Urban', 'Rural'], n, p=[0.65, 0.35]),
    
    # Variabile principale
    'Tip_Cafea_Preferat': np.random.choice(
        ['Espresso', 'Latte', 'Americano', 'Cappucino', 'Decaf', 'Alte'], 
        n, 
        p=[0.25, 0.20, 0.28, 0.20, 0.05, 0.02]
    ),
    
    'Numar_Cani_Pe_Zi': np.random.choice(
        [1, 2, 3, 4, 5], 
        n, 
        p=[0.20, 0.35, 0.25, 0.15, 0.05]
    ),
    
    'Ora_Consum_Dominanta': np.random.choice(
        ['Dimineata', 'Pranz', 'Seara'], 
        n, 
        p=[0.65, 0.25, 0.10]
    ),
    
    'Buget_Luna_MDL': np.random.randint(200, 2000, n),
    
    'Loc_Achizitie': np.random.choice(
        ['Cafenea', 'Supermarket', 'Oficiu', 'Distributie', 'Acasa', 'Alte'],
        n,
        p=[0.25, 0.20, 0.20, 0.20, 0.10, 0.05]
    ),
}
```

```python
# Creare DataFrame
df = pd.DataFrame(data)

# Ajustări logice pentru coerență
# Persoanele mai tinere tind să consume mai mult latte
df.loc[(df['Varsta'] < 30) & (np.random.rand(n) < 0.4), 'Tip_Cafea_Preferat'] = 'Latte'

# Persoanele mai în vârstă tind să consume mai mult decaf
df.loc[(df['Varsta'] > 55) & (np.random.rand(n) < 0.3), 'Tip_Cafea_Preferat'] = 'Decaf'

# Urban tinde să aibă buget mai mare
urban_mask = df['Tip_Localitate'] == 'Urban'
df.loc[urban_mask, 'Buget_Luna_MDL'] = (
    df.loc[urban_mask, 'Buget_Luna_MDL'] * np.random.uniform(1.1, 1.3, sum(urban_mask))).astype(int)

df.head(5)
```

```python
# Salvare în CSV
df.to_csv('./dataset/date_cafea_consum.csv', index=False, encoding='utf-8-sig')

print("✓ Fișier CSV generat cu succes: date_cafea_consum.csv")
print(f"\n📊 Statistici generale:")
print(f"   - Total înregistrări: {len(df)}")
print(f"   - Vârstă medie: {df['Varsta'].mean():.1f} ani")
print(f"   - Consum mediu zilnic: {df['Numar_Cani_Pe_Zi'].mean():.2f} căni")
print(f"   - Buget mediu lunar: {df['Buget_Luna_MDL'].mean():.2f} MDL")
print(f"\n📋 Primele 5 înregistrări:")
# print example of table
# pandas settings
pd.set_option('display.max_rows', 500)
pd.set_option('display.max_columns', 500)
pd.set_option('display.width', 1000)
pd.options.display.width = 0
df.head(5)
```
